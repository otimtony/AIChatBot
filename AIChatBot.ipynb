```python
import os
import gradio as gr
from openai import OpenAI
from dotenv import load_dotenv
```

```python
# Load OPENAI API KEY
load_dotenv()

api_key = os.getenv("NEW_OPENAI_API_KEY")

if not api_key:
    raise ValueError(
        "OPENAI_API_KEY was not found. "
        "Please create a .env file and add your API key."
    )

client = OpenAI(api_key=api_key)

print("OpenAI API configuration loaded successfully.", api_key)
```

```python
# CHAT SYSTEM PROMPT

SYSTEM_PROMPT = """
You are StudyMate, an AI Academic Advisor designed to help
college and university students.

Your main responsibilities are:

1. Explain academic concepts clearly and accurately.
2. Adapt explanations to the student's knowledge level.
3. Provide examples when they improve understanding.
4. Help students create realistic study plans.
5. Generate practice questions and quizzes when requested.
6. Break difficult topics into smaller steps.
7. Encourage students to understand concepts instead of simply
   copying answers.
8. Ask a clarifying question when a student's request is too vague.
9. Stay focused on academic and educational assistance.

Response guidelines:

- Use clear and simple language.
- Use bullet points or numbered lists when appropriate.
- Give practical examples for technical concepts.
- For study plans, consider the student's available time.
- Explain technical concepts step-by-step.
- Do not invent university policies, deadlines, grades, or
  course requirements.
- If official university information is required, tell the student
  to verify it with their university.
- If you are uncertain about information, clearly say so.
- Keep responses focused and useful.
"""
```

```python
# Input Validation

def validate_input(message):

    if message is None:
        return False, "Please enter a question."

    message = message.strip()

    if len(message) < 3:
        return False, "Please enter a more detailed question."

    if len(message) > 2000:
        return False, (
            "Your question is too long. "
            "Please keep it under 2,000 characters."
        )

    return True, message
```

```python
# TEST INPUT VALIDATION

test_inputs = [
    "",
    "Hi",
    "Explain neural networks."
]

for test in test_inputs:

    valid, result = validate_input(test)

    print(f"Input: {test!r}")
    print(f"Valid: {valid}")
    print(f"Result: {result}")
    print("-" * 50)
```

```python
# Main Chatbot Function

def academic_advisor(message, history):

    #Validate the user's input
    valid, result = validate_input(message)

    # If validation fails, return the error message
    if not valid:
        return result

    #Start the message list
    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        }
    ]

    #Add previous conversation
    for user_message, assistant_message in history:

        messages.append(
            {
                "role": "user",
                "content": user_message
            }
        )

        messages.append(
            {
                "role": "assistant",
                "content": assistant_message
            }
        )

    # Add the current user question
    messages.append(
        {
            "role": "user",
            "content": result
        }
    )

    # Send request to the LLM
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        temperature=0.4
    )

    #Extract the generated response
    answer = response.choices[0].message.content

    # Return the answer to Gradio
    return answer
```

```python
# Test the Chatbot Backend

test_question = "Explain supervised learning in simple terms."

# Start with an empty conversation
test_history = []

# Send the question to the chatbot
response = academic_advisor(
    test_question,
    test_history
)

# Display the response
print("Student:")
print(test_question)

print("\nStudyMate:")
print(response)
```
